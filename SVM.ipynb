```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
```

```python
for i in df.columns:
  le=LabelEncoder()
  if df[i].dtype =="object":
    df[i] = le.fit_transform(df[i])
```

```python
df.head()
```

```python
l1 = ['Sex',"Drug"]
l2 = ["BP","Cholesterol"]
```

```python
for i in l1:
  df[i]=le.fit_transform(df[i])
```

```python
oe = OrdinalEncoder(categories=[["LOW","NORMAL","HIGH"],["NORMAL","HIGH"]])
```

```python
df[l2] = oe.fit_transform(df[l2])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop('Drug',axis=1),df['Drug'],test_size=0.2,random_state=42)
```

```python
sns.scatterplot(data=df,x="BP",y="Sex",)
```

```python
from sklearn.svm import SVC
```

```python
model = SVC(kernel="rbf")
```

```python
model.fit(x_train,y_train)
```

```python
result = pd.DataFrame()
result["Actual"] = y_test
ypred = model.predict(x_test)
result["Predicted"] = ypred
```
