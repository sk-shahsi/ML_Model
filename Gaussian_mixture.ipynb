```python
import numpy as np
from sklearn.datasets import make_blobs
```

```python
x,y=make_blobs(n_samples=500, centers=3,cluster_std=[1.0,2.5,0.5],random_state=42)
```

```python
x
```

```python
y
```

```python
# kmens
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

```

```python
kmeans = KMeans(n_clusters=3,random_state=42)
```

```python
kmeans_lable = kmeans.fit_predict(x)
```

```python
kmeans_lable
```

```python
gmm = GaussianMixture(n_components=3,random_state=42)
gmm_lable = gmm.fit_predict(x)
```

```python
import matplotlib.pyplot as plt
```

```python
fig,axes = plt.subplots(1,3,figsize=(15,5))

axes[0].scatter(x[:,0],x[:,1],c=y,s=30)
axes[0].set_title(' Truth Cluster')




axes[2].scatter(x[:,0],x[:,1],c=gmm_lable,s=30)
axes[2].set_title('GMM Clustring')

axes[1].scatter(x[:,0],x[:,1],c=kmeans_lable,s=30)
axes[1].set_title('K-Means')

plt.show()
```
