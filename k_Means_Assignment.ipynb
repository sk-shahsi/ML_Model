```python
df = df.drop(['status_id','status_published'],axis=1)
```

```python
df = df.drop(['Column1','Column2','Column3','Column4'],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le =LabelEncoder()
```

```python
df['status_type'] = le.fit_transform(df['status_type'])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
X=scaler.fit_transform(df)
```

```python
from sklearn.cluster import KMeans
```

```python
k = KMeans(n_clusters=3, random_state=42)
k.fit(X)
```

```python
df['Cluster'] = k.labels_
```

```python
print(df.head())
```

```python
centers = k.cluster_centers_
print(centers)
```

```python
overall_mean = X.mean(axis=0)
```

```python
cluster_counts = np.bincount(k.labels_)
```

```python
inter_variance = 0

for i in range(2):
    inter_variance += cluster_counts[i] * np.sum(
        (centers[i] - overall_mean) ** 2
    )

print("Inter-cluster variance:", inter_variance)
```
