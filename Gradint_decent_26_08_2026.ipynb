#### Gradient Desent -> Gradient descent is an iterative optimization algorithm used to find the minimum of a function. In machine learning, it minimizes error (loss) by adjusting model parameters (like weights and biases) step-by-step in the direction of the steepest decrease.

```python
# 1. Calculate the slope
# 2. Determine which direction goes downhill
# 3. Take small step (learning rate)
# 4. Calculate the Slope again
# 5. Repeat
```

```python
# change in loss with respect to change in weight
```

```python
# d loss / d w
```

```python
# If gradint comes as positive you need to decrease the weight / bias
# If gradint comes as negative you need to increase the weight / bias
```

```python
# Types of GD ->
```

```python
# 1. Batch Gradient Descent ->
# Uses all the training data to calculate one gradient
# Uses when you have small datasets

# 1 Update = Entire Dataset
```

```python
# 2. Stochastic Gradient Descent ->
# Use one sample at a time
# Sample 1 -> Update
# Sample 2 -> Update
# Sample 3 -> Update
# ...

# 1 Update = 1 Sample
# It is noisy because every individual sample can give a slightly different gradient
```

```python
# 3. Mini Batch Gradient Descent ->
# Uses a small group of samples

# 1000 Samples

# 32 Samples -> Update
# 32 Samples -> Update
# 32 Samples -> Update
# ...

# 1 Update = Small Batch

# This is the most commonly used approach in Deep Learning
```

```python
# def initialize()

# def predict_y() -> y = wX + B

# def get_cost()

# def update_weight_bias()

# def run_gradient_descent()
```

```python
import random
```

```python
def initialized(dim):
  b = random.random()           # bias
  theta = np.random.rand(dim)   # weight
  return b, theta
# y = w1x1 + w2x2 + w3x3 + .... + B

b, theta = initialized(2)
print(b, theta)
```

```python
def predict_Y(b, theta, X):
  return b + np.dot(X, theta)

Y_hat = predict_Y(b, theta, X)      # Prediction
```

```python
from sklearn.metrics import *
```

```python
from sklearn.metrics import *
def get_cost(y, Y_hat):
  return mean_squared_error(y, Y_hat)

get_cost(y, Y_hat)




```

```python
def update_theta(x,y,y_hat,b_0, theta_0, learning_rate):
  db = (np.sum(y_hat-y)*2)/len(y)     # d(Loss)/d(B)
  dw = (np.dot((y_hat-y), x)*2)/len(y)    # d(Loss)/d(W)
  b_1 = b_0 - learning_rate*db
  theta_1 = theta_0 - learning_rate*dw
  return b_1, theta_1

b, theta = update_theta(X, y, Y_hat, b, theta, 0.01)

print(b, theta)
```

```python
def run_gradient_descent(X, Y, alpha, num_iterations):
  b, theta = initialized(X.shape[1])
  ite_num = 0
  gd_iterations_df = pd.DataFrame(columns = ["Iteration", "Cost"])
  result_idx = 0
  for i in range(num_iterations):
    Y_hat = predict_Y(b, theta, X)
    this_cost = get_cost(Y, Y_hat)
    prev_b = b
    prev_theta = theta
    b, theta = update_theta(X, Y, Y_hat, prev_b, prev_theta, alpha)
    # Store the iteration and cost
    if(ite_num%10 == 0):
      gd_iterations_df.loc[result_idx] = [ite_num, this_cost]
      result_idx = result_idx+1
    ite_num = ite_num+1
  return gd_iterations_df, b, theta

gd_iteration_df, b, theta = run_gradient_descent(X, y, alpha=0.01, num_iterations=1000)
```
