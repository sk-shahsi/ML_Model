OLS -> It is a technique used to find the best fit line(hyperplane) in Linear Regression, by minimizing the sum of squared distances(errors) between the actual and predicted data points
Least Squares -> It minimizes he squared difference between the actual and predicted values
OLS is used to get the coffecient for the relation between Input features and the Target columns
Coffiecient -> m -> Slope, c -> intercept

```python
# How OLS works ->
# B = [(X.T @ X)^-1] @ X.T @ y
# X.T -> Transpose matrix of X
# ^ -1 -> Inverse
# @ -> Dot product
# X -> Input Feature
# y -> Output Feature
# B -> Coffecient
```

```python
x=df['YearsExperience'].values
y=df['Salary'].values
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.1,random_state=42)
```

```python
x_train = pd.DataFrame(x_train)
x_test = pd.DataFrame(x_test)
y_train = pd.DataFrame(y_train)
y_test = pd.DataFrame(y_test)
```

```python
import numpy as np
```

```python
x_train_ols = np.hstack((np.ones((x_train.shape[0], 1)), x_train))
x_test_ols = np.hstack((np.ones((x_test.shape[0], 1)), x_test))
```

```python
x_train_ols
```

```python
x_tanspose_ols = x_train_ols.T
```

```python
beta = np.linalg.inv(x_tanspose_ols @ x_train_ols) @ x_tanspose_ols @ y_train
```

```python
y_pred_ols = x_test_ols @ beta
```

```python
y_pred_ols
```

```python
y_test
```

```python
r2_score(y_test, y_pred_ols)
```

```python
mean_squared_error(y_test, y_pred_ols)
```
