```python
# What is Markov Chain->

# A Markov Chain is a sequence of event where the outcome of the next event depend onaly on the current state,not on the sequence  of event that preceded it.


#This is called the Memoryless Property
```

```python
# Example->

# Predicting the Weather->

# If today is sunny, tomorrow is likely to be
# sunny 80%
# Rainy 10%
# Cloudy 10%

# There probablity form a system where the future depends onaly on the prsent weather, not on what happened yested,earlier
```

```python
# Key Components->

# 1.State->
# The possibal condition or outcome in the system(eg,Sunny,Ranny,Cloudy)

# 2. Transiction Probablities->
# The likehood of moving from one state to another.represent between 0 to 1

# 3 Transitionaing Matrix->

# A Table showing the probablities of transitionaing between state
```

```python
# How dose markkov Chain Works??

# 10) Starting State->
# Being with an initial state(eg.Sunny

# 2) Transition->

# Use probablities to determine the next state

# From example , if the current state is sunny:
#Stay Sunnny -> 80%

#Transition to Ranny ->10%

#Transition to Cloudy ->10%

# 3) Repete the process:
# Keep transitioning fro the desired number of steps(eg .next 10 days)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Defining the state
state = ["sunny","rany","cloudy"]

#Define the transion matrix

transition_matrix = [[0.8,0.1,0.1],
                     [0.2,0.6,0.2],
                      [0.3,0.3,0.4]]


def simulated_weather(n_days):
  current_state = 0
  weather_sequence = [state[current_state]]
  for i in range (0, n_days):
    current_state = np.random.choice([0,1,2])
    weather_sequence.append(state[current_state])
  return weather_sequence

simulated_weather(5)
```
