```python
from sklearn.preprocessing import StandardScaler
scalear = StandardScaler()

```

```python
x_scaled = scalear.fit_transform(x)
```

```python
x_scaled
```

```python
x_scaled.T
```

```python
cov_mat = np.cov(x_scaled.T)
pd.DataFrame(cov_mat)
```

```python
eig_val, eig_vec = np.linalg.eig(cov_mat)
```

```python
pd.DataFrame(eig_val)
```

```python
pd.DataFrame(eig_vec)
```

```python
total = sum(eig_val)
```

```python
exp_var = []
for i in sorted(eig_val,reverse=True):
  exp_var.append((i/total)*100)
exp_var
```

```python
cum_exp_var = np.cumsum(exp_var)
pd.DataFrame(cum_exp_var)
```

```python
cum_exp_var = np.cumsum(exp_var)

df = pd.DataFrame(cum_exp_var, columns=["Cumulative Explained Variance"])
df.plot(kind='bar')
plt.xlabel("Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.show()
```
