```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/Salary_dataset.csv')
```

```python
x = df['YearsExperience']
y= df['Salary']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train = pd.DataFrame(x_train)
x_test = pd.DataFrame(x_test)
y_train = pd.DataFrame(y_train)
y_test = pd.DataFrame(y_test)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
r2_score(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
#OLS
```

```python
x1=df[["YearsExperience"]].values
y1=df["Salary"].values
```

```python
x1_train,x1_test,y1_test,y1_train=train_test_split(x1,y1,test_size=0.2,random_state=42)
```

```python
import numpy as np
```

```python
x_train_ols = np.hstack((np.ones((x_train.shape[0],1)),x_train))
x_test_ols = np.hstack((np.ones((x_test.shape[0],1)),x_test))
```

```python
x_train_ols
```

```python
x_transpose_ols = x_train_ols.T
```

```python
x_transpose_ols
```

```python
beta = np.linalg.inv(x_transpose_ols @ x_train_ols) @ x_transpose_ols @ y_train
```

```python
y_pred_ols = x_test_ols @ beta
```

```python
y_pred_ols
```

```python
y1_test
```

```python
r2_score(y1_test,y_pred_ols)
```

```python
mean_squared_error(y1_test,y_pred_ols)
```
