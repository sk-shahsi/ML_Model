```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#
```

```python
df = pd.read_csv("/content/nasa.csv")
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.replace([" ","?"],np.nan,inplace=True)
```

```python
df.info()
```

```python

df["Close Approach Date"] = pd.to_datetime(df["Close Approach Date"])
df["Orbit Determination Date"] = pd.to_datetime(df["Orbit Determination Date"])
```

```python
df.info()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df.corr(numeric_only=True),cmap="YlOrRd")
plt.show()
```

```python
df = df.drop(['Est Dia in M(min)', 'Est Dia in M(max)',
       'Est Dia in Miles(min)', 'Est Dia in Miles(max)',
       'Est Dia in Feet(min)', 'Est Dia in Feet(max)', 'Relative Velocity km per hr', 'Miss Dist.(lunar)',
       'Miss Dist.(kilometers)', 'Miss Dist.(miles)', 'Close Approach Date', 'Orbit Determination Date', 'Neo Reference ID', 'Name',
              "Equinox", 'Orbiting Body'],axis=1)
```

```python
df['Hazardous'].value_counts()
```

```python
sns.countplot(data=df,x='Hazardous')
plt.show()
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
x=df.drop('Hazardous',axis=1)
y=df['Hazardous']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=GaussianNB()
model.fit(x_train,y_train)
```

```python
#Test model
y_pred=model.predict(x_test)
```

```python
#evaluate
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy)
```

```python
# confusionmatrix
cm=confusion_matrix(y_test,y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
print("Classification Report:")
print(classification_report(y_test,y_pred))
```
